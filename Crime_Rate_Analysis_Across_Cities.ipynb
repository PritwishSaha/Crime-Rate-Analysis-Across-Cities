```python
import pandas as pd
df = pd.read_csv('crime_dataset_india.csv')
df.head()
```

```python
date_columns = [
    'Date Reported',
    'Date of Occurrence',
    'Time of Occurrence',
    'Date Case Closed'
]
for col in date_columns:
    df[col] = pd.to_datetime(
        df[col],
        format='%d-%m-%Y %H:%M',
        errors='coerce'
    )
```

```python
df['Report Number'].duplicated().sum()
```

```python
df.duplicated().sum()
```

```python
df['Weapon Used'] = df['Weapon Used'].fillna('Unknown')
```

```python
pd.crosstab(
    df['Case Closed'],
    df['Date Case Closed'].isna()
)
```

```python
df['Year'] = df['Date of Occurrence'].dt.year
```

```python
df['Month'] = df['Date of Occurrence'].dt.month
```

```python
df['Month Name'] = df['Date of Occurrence'].dt.month_name()
```

```python
df['Day of Week'] = df['Date of Occurrence'].dt.day_name()
```

```python
df['Hour'] = df['Time of Occurrence'].dt.hour
```

```python
def get_time_period(hour):
    if 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'
df['Time Period'] = df['Hour'].apply(get_time_period)
```

```python
df['Case Closure Days'] = (
    df['Date Case Closed'] - df['Date Reported']
).dt.total_seconds() / (24 * 60 * 60)
```
